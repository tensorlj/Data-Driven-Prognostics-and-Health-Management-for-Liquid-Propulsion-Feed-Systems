```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN
from sklearn.neighbors import NearestNeighbors, KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, precision_score, recall_score

# ------------------------------------------------------------------
# Input: feat_df from the FeatureEngineering class
#   (one row per window: recording_id, window, label, label_name, then the features)
# ------------------------------------------------------------------
META_COLS = ["recording_id", "window", "label", "label_name"]
SEED = 42
TASK_TITLE = "Normal vs abnormal (valve-opening windows)"

# ---- 1) Split by RECORDING: 60 / 20 / 20, stratified by label ----
rec_labels = feat_df.groupby("recording_id")["label"].first()
ids, y = rec_labels.index.to_numpy(), rec_labels.to_numpy()

ids_train, ids_tmp, y_train, y_tmp = train_test_split(
    ids, y, test_size=0.40, stratify=y, random_state=SEED)
ids_val, ids_test, y_val, y_test = train_test_split(
    ids_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=SEED)

split = {"train": [int(i) for i in ids_train],
         "val":   [int(i) for i in ids_val],
         "test":  [int(i) for i in ids_test]}

# leakage check: no recording in more than one set
assert not (set(split["train"]) & set(split["val"]))
assert not (set(split["train"]) & set(split["test"]))
assert not (set(split["val"])   & set(split["test"]))
with open("split.json", "w") as f:
    json.dump(split, f)

lab = rec_labels.to_dict()
for s, idx in split.items():
    print(f"{s:5s}: {len(idx):3d} recordings | normal={sum(lab[i]==0 for i in idx)}, "
          f"abnormal={sum(lab[i]==1 for i in idx)}")

# ---- 2) Windows of each split (all windows of a recording stay in its own split) ----
F = {s: feat_df[feat_df["recording_id"].isin(split[s])].reset_index(drop=True) for s in split}

# drop constant features (judged on train), standardize (fit on train only)
fcols = [c for c in feat_df.columns if c not in META_COLS]
keep = F["train"][fcols].to_numpy(float).std(axis=0) > 1e-12
fcols = [c for c, k in zip(fcols, keep) if k]
scaler = StandardScaler().fit(F["train"][fcols])
X    = {s: scaler.transform(F[s][fcols]) for s in F}
meta = {s: F[s][["recording_id", "window", "label"]] for s in F}
ytr  = F["train"]["label"].to_numpy()

print(f"\n{len(fcols)} features | windows: " + ", ".join(f"{s}={len(F[s])}" for s in F) +
      f" | normal train windows={(ytr==0).sum()}")

# ------------------------------------------------------------------
def thr_3sigma(s):
    return s.mean() + 3 * s.std()

def make_unsupervised(Xn, k_clusters=3, db_minpts=5):
    """Fit each model on normal windows only.
    Returns {name: (score_fn, train_scores)}; larger score = more abnormal."""
    models = {}

    # PCA: reconstruction error (95% variance)
    pca = PCA(n_components=0.95, svd_solver="full").fit(Xn)
    def pca_score(X):
        R = pca.inverse_transform(pca.transform(X))
        return ((X - R) ** 2).sum(axis=1)
    models["PCA"] = (pca_score, pca_score(Xn))

    # K-Means: distance to nearest cluster centre
    km = KMeans(n_clusters=k_clusters, n_init=10, random_state=0).fit(Xn)
    km_score = lambda X: km.transform(X).min(axis=1)
    models[f"KMeans(k={k_clusters})"] = (km_score, km_score(Xn))

    # DBSCAN: eps = 90th percentile of k-distances; score = distance to nearest core point
    nn_k = NearestNeighbors(n_neighbors=db_minpts).fit(Xn)
    kdist = nn_k.kneighbors(Xn)[0][:, -1]
    eps = np.percentile(kdist, 90)
    db = DBSCAN(eps=eps, min_samples=db_minpts).fit(Xn)
    core_idx = db.core_sample_indices_
    assert len(core_idx) >= 2, "DBSCAN found <2 core points; increase eps percentile"
    nn_core = NearestNeighbors(n_neighbors=2).fit(Xn[core_idx])
    def db_score(X):
        return nn_core.kneighbors(X, n_neighbors=1)[0][:, 0]
    d2 = nn_core.kneighbors(Xn, n_neighbors=2)[0]
    is_core = np.zeros(len(Xn), dtype=bool); is_core[core_idx] = True
    db_train = np.where(is_core, d2[:, 1], d2[:, 0])    # leave-self-out for core points
    models["DBSCAN"] = (db_score, db_train)

    return models

def make_supervised():
    return {
        "DecisionTree": DecisionTreeClassifier(max_depth=5, class_weight="balanced", random_state=0),
        "LogReg":       LogisticRegression(max_iter=1000, class_weight="balanced", C=10.0),
        "NaiveBayes":   GaussianNB(priors=[0.5, 0.5]),
        "KNN(k=3)":     KNeighborsClassifier(n_neighbors=3),
        "ANN":          MLPClassifier(hidden_layer_sizes=(64, 32), activation="relu",
                                      alpha=1e-3, max_iter=1000, random_state=0),
        "RandomForest": RandomForestClassifier(n_estimators=300, class_weight="balanced",
                                               random_state=0, n_jobs=-1),
        "GradBoost":    GradientBoostingClassifier(n_estimators=100, learning_rate=0.1,
                                                   max_depth=3, random_state=0),
    }

SUPERVISED = set(make_supervised().keys())      # always matches the model names above

# ------------------------------------------------------------------
# Evaluation: accuracy, F1 (abnormal = positive class), confusion matrix
# at window level and at recording level (majority vote of the 3 windows)
# ------------------------------------------------------------------
def evaluate(meta, pred):
    """Recording level: each recording's label = majority vote of its windows."""
    m = meta[["recording_id", "label"]].copy()
    m["pred"] = pred
    rec = m.groupby("recording_id").agg(label=("label", "first"), vote=("pred", "mean"))
    yt = rec["label"].to_numpy()
    yp = (rec["vote"] > 0.5).astype(int).to_numpy()
    return {"rec": {
        "acc":       accuracy_score(yt, yp),
        "f1":        f1_score(yt, yp, pos_label=1, zero_division=0),
        "precision": precision_score(yt, yp, pos_label=1, zero_division=0),
        "recall":    recall_score(yt, yp, pos_label=1, zero_division=0),
        "cm":        confusion_matrix(yt, yp, labels=[0, 1]),   # rows = true, cols = predicted
        "n":         len(yt),
    }}

def metrics_table(results):
    rows = []
    for name, r in results.items():
        rows.append({
            "model": name,
            "type": "supervised" if name in SUPERVISED else "unsupervised",
            "val_acc%":   round(100 * r["val"]["rec"]["acc"], 1),
            "val_F1":     round(r["val"]["rec"]["f1"], 3),
            "test_acc%":  round(100 * r["test"]["rec"]["acc"], 1),
            "test_F1":    round(r["test"]["rec"]["f1"], 3),
            "test_precision": round(r["test"]["rec"]["precision"], 3),
            "test_recall":    round(r["test"]["rec"]["recall"], 3),
        })
    return pd.DataFrame(rows).set_index("model")

def plot_confusions(results, split_name, level="rec", ncol=3):
    names = list(results.keys())
    nrow = int(np.ceil(len(names) / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(4.2 * ncol, 3.8 * nrow))
    axes = np.atleast_1d(axes).ravel()
    for ax, nm in zip(axes, names):
        r = results[nm][split_name][level]
        cm = r["cm"]
        ax.imshow(cm, cmap="Blues", vmin=0)
        for i in range(2):
            for j in range(2):
                ax.text(j, i, int(cm[i, j]), ha="center", va="center", fontsize=15,
                        color="white" if cm[i, j] > cm.max() / 2 else "black")
        ax.set_xticks([0, 1]); ax.set_xticklabels(["Normal", "Abnormal"])
        ax.set_yticks([0, 1]); ax.set_yticklabels(["Normal", "Abnormal"])
        ax.set_xlabel("Predicted"); ax.set_ylabel("True")
        ax.set_title(f"{nm}\nAcc={100*r['acc']:.1f}%  F1={r['f1']:.3f}", fontsize=10)
    for ax in axes[len(names):]:
        ax.axis("off")
    lvl_name = "recording level" if level == "rec" else "window level"
    fig.suptitle(f"{TASK_TITLE} - confusion matrices ({split_name}, {lvl_name})", fontsize=13)
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------------
# Fit every model and evaluate
# ------------------------------------------------------------------
preds = {}
# unsupervised: fit on normal windows only, threshold = mean + 3 sigma
for name, (fn, tr_scores) in make_unsupervised(X["train"][ytr == 0]).items():
    thr = thr_3sigma(tr_scores)
    preds[name] = {s: (fn(X[s]) > thr).astype(int) for s in ["val", "test"]}
# supervised: fit on all training windows with labels
for name, clf in make_supervised().items():
    clf.fit(X["train"], ytr)
    preds[name] = {s: clf.predict(X[s]) for s in ["val", "test"]}

results = {name: {s: evaluate(meta[s], p[s]) for s in ["val", "test"]}
           for name, p in preds.items()}

print(f"\n===== {TASK_TITLE} =====")
print("Recording level, abnormal = positive class (precision and recall: test only):")
display(metrics_table(results))
plot_confusions(results, "val",  "rec")
plot_confusions(results, "test", "rec")
# ---- compact summary: recording level, validation vs test ----
summary = pd.DataFrame({
    ("accuracy %", "val"):  {n: 100 * r["val"]["rec"]["acc"]  for n, r in results.items()},
    ("accuracy %", "test"): {n: 100 * r["test"]["rec"]["acc"] for n, r in results.items()},
    ("F1",         "val"):  {n: r["val"]["rec"]["f1"]  for n, r in results.items()},
    ("F1",         "test"): {n: r["test"]["rec"]["f1"] for n, r in results.items()},
}).round(3)
print("\n===== Recording-level summary =====")
display(summary)
```

```python
# ------------------------------------------------------------------
# Task 2: fault type of the abnormal recordings  (Bubble vs SV fault)
# Needs from earlier cells: pipe, feat_df, split, F, META_COLS, make_unsupervised,
#   make_supervised, thr_3sigma, SUPERVISED, evaluate, metrics_table
# Positive class (1) = SV fault, 0 = Bubble
# ------------------------------------------------------------------
TASK2_TITLE = "Task 2: Bubble vs SV fault (abnormal recordings)"
TYPE_NAMES = ["Bubble", "SV fault"]

if pipe.body is None:                      # only the saved CSV was loaded: read the labels table
    pipe.load(verbose=False)

# ---- Task 2 labels: Fault = SV fault (1), Anomaly = Bubble (0) ----
cond = pipe.body.set_index("case")["condition"].str.lower()
task2_type = {int(c): (1 if v == "fault" else 0) for c, v in cond.items() if v != "normal"}

print("Abnormal recordings per type:")
for s in split:
    t = [task2_type[i] for i in split[s] if i in task2_type]
    print(f"  {s:5s}: Bubble={t.count(0):3d} | SV fault={t.count(1):3d}")

# ---- Abnormal windows of each split (same recording split as Task 1) ----
F2 = {}
for s in split:
    d = F[s][F[s]["recording_id"].isin(task2_type)].copy()
    d["label"] = d["recording_id"].map(task2_type).astype(int)     # label is now the fault type
    F2[s] = d.reset_index(drop=True)

# drop constant features (judged on train), standardize (fit on train only)
fcols2 = [c for c in feat_df.columns if c not in META_COLS]
keep = F2["train"][fcols2].to_numpy(float).std(axis=0) > 1e-12
fcols2 = [c for c, k in zip(fcols2, keep) if k]
scaler2 = StandardScaler().fit(F2["train"][fcols2])
X2    = {s: scaler2.transform(F2[s][fcols2]) for s in F2}
meta2 = {s: F2[s][["recording_id", "window", "label"]] for s in F2}
ytr2  = F2["train"]["label"].to_numpy()

print(f"\n{len(fcols2)} features | windows: " + ", ".join(f"{s}={len(F2[s])}" for s in F2) +
      f" | Bubble train windows={(ytr2 == 0).sum()}")

# ---- Fit every model ----
preds2 = {}
# unsupervised: fit on Bubble windows only, SV fault = outside mean + 3 sigma
for name, (fn, tr_scores) in make_unsupervised(X2["train"][ytr2 == 0]).items():
    thr = thr_3sigma(tr_scores)
    preds2[name] = {s: (fn(X2[s]) > thr).astype(int) for s in ["val", "test"]}
# supervised: fit on all abnormal training windows with their fault type
for name, clf in make_supervised().items():
    clf.fit(X2["train"], ytr2)
    preds2[name] = {s: clf.predict(X2[s]) for s in ["val", "test"]}

results2 = {name: {s: evaluate(meta2[s], p[s]) for s in ["val", "test"]}
            for name, p in preds2.items()}

# ---- Results table (same layout as Task 1) ----
print(f"\n===== {TASK2_TITLE} =====")
print("Recording level, SV fault = positive class (precision and recall: test only):")
display(metrics_table(results2))
```

## Task-3

```python
# ------------------------------------------------------------------
# Task 3: bubble location (8 classes: BP1-BP7, BV1)
# Needs from earlier cells: pipe, feat_df, META_COLS, make_supervised, metrics_table
# ------------------------------------------------------------------
TASK3_TITLE = "Task 3: Bubble location (bubble recordings)"
LOC_NAMES = ["BP1", "BP2", "BP3", "BP4", "BP5", "BP6", "BP7", "BV1"]
SEED3 = 42

# ---- Location labels from the full labels table ----
raw = pd.read_excel(pipe.labels_path, header=None)
norm = raw.astype(str).apply(lambda col: col.str.strip().str.lower())
hdr_row = norm.index[(norm == "condition").any(axis=1)][0]

cols = ["case", "spacecraft", "condition", "SV1", "SV2", "SV3", "SV4",
        "BP1", "BP2", "BP3", "BP4", "BP5", "BP6", "BP7", "BV1"]
lab_full = raw.loc[hdr_row + 1:, :].iloc[:, :15].copy()
lab_full.columns = cols
lab_full["case"] = pd.to_numeric(lab_full["case"], errors="coerce")
lab_full = lab_full.dropna(subset=["case"]).copy()
lab_full["case"] = lab_full["case"].astype(int)

bub = lab_full[lab_full["condition"].astype(str).str.strip().str.lower() == "anomaly"].copy()
is_yes = bub[LOC_NAMES].astype(str).apply(lambda c: c.str.strip().str.lower() == "yes")
bub["location"] = is_yes.apply(lambda r: "+".join(r.index[r]), axis=1)
print("Bubble recordings:", len(bub))
print(bub["location"].value_counts().to_string())

# bubble recording -> location index (0..7)
assert set(bub["location"]) <= set(LOC_NAMES), "unexpected location labels"
loc_of = {int(c): LOC_NAMES.index(l) for c, l in zip(bub["case"], bub["location"])}
assert all(i in set(feat_df["recording_id"]) for i in loc_of), "some bubble recordings are not in feat_df"

# ---- Split by recording: 1 train / 1 val / 1 test per location ----
rng = np.random.RandomState(SEED3)
split3 = {"train": [], "val": [], "test": []}
for k in range(len(LOC_NAMES)):
    ids_k = sorted(i for i, v in loc_of.items() if v == k)
    assert len(ids_k) == 3, f"{LOC_NAMES[k]} has {len(ids_k)} recordings"
    rng.shuffle(ids_k)
    for s, i in zip(["train", "val", "test"], ids_k):
        split3[s].append(int(i))

assert not (set(split3["train"]) & set(split3["val"]))
assert not (set(split3["train"]) & set(split3["test"]))
assert not (set(split3["val"])   & set(split3["test"]))
with open("split_bubble.json", "w") as f:
    json.dump(split3, f)
print()
for s, ids_s in split3.items():
    print(f"{s:5s}: {len(ids_s)} recordings -> {sorted(ids_s)}")

# ---- Windows of each split, taken from feat_df ----
F3 = {}
for s in split3:
    d = feat_df[feat_df["recording_id"].isin(split3[s])].copy()
    d["label"] = d["recording_id"].map(loc_of).astype(int)          # label is now the location
    F3[s] = d.reset_index(drop=True)

# drop constant features (judged on train), standardize (fit on train only)
fcols3 = [c for c in feat_df.columns if c not in META_COLS]
keep = F3["train"][fcols3].to_numpy(float).std(axis=0) > 1e-12
fcols3 = [c for c, k in zip(fcols3, keep) if k]
scaler3 = StandardScaler().fit(F3["train"][fcols3])
X3    = {s: scaler3.transform(F3[s][fcols3]) for s in F3}
meta3 = {s: F3[s][["recording_id", "window", "label"]] for s in F3}
ytr3  = F3["train"]["label"].to_numpy()

print(f"\n{len(fcols3)} features | windows: " + ", ".join(f"{s}={len(F3[s])}" for s in F3))

# ---- Evaluation: recording level (majority vote of its windows), macro over the 8 locations ----
def eval_loc(meta, pred):
    m = meta[["recording_id", "label"]].copy()
    m["pred"] = np.asarray(pred).astype(int)
    rec = m.groupby("recording_id").agg(
        label=("label", "first"),
        vote=("pred", lambda p: int(np.bincount(p, minlength=8).argmax())))
    yt, yp = rec["label"].to_numpy(), rec["vote"].to_numpy()
    L = list(range(len(LOC_NAMES)))
    return {"rec": {
        "acc":       accuracy_score(yt, yp),
        "f1":        f1_score(yt, yp, labels=L, average="macro", zero_division=0),
        "precision": precision_score(yt, yp, labels=L, average="macro", zero_division=0),
        "recall":    recall_score(yt, yp, labels=L, average="macro", zero_division=0),
        "cm":        confusion_matrix(yt, yp, labels=L),     # rows = true, cols = predicted
        "n":         len(yt),
    }}

# ---- Fit the supervised classifiers ----
models3 = make_supervised()
models3["NaiveBayes"] = GaussianNB(priors=[1 / len(LOC_NAMES)] * len(LOC_NAMES))   # equal priors, 8 classes
results3 = {}
for name, clf in models3.items():
    clf.fit(X3["train"], ytr3)
    results3[name] = {s: eval_loc(meta3[s], clf.predict(X3[s])) for s in ["val", "test"]}

# ---- Results table (same layout as Task 1) ----
print(f"\n===== {TASK3_TITLE} =====")
print("Recording level, F1 / precision / recall are macro averages over the 8 locations "
      "(precision and recall: test only):")
display(metrics_table(results3))
```

## Task-4

```python
# ------------------------------------------------------------------
# Task 4: which solenoid valve failed (4 classes: SV1-SV4)
# Needs from earlier cells: pipe, feat_df, META_COLS, make_supervised, metrics_table
# ------------------------------------------------------------------
TASK4_TITLE = "Task 4: Failed valve (SV-fault recordings)"
VALVE_NAMES = ["SV1", "SV2", "SV3", "SV4"]
NCLS = len(VALVE_NAMES)
SEED4 = 42

# ---- SV-fault table from the full labels file ----
raw = pd.read_excel(pipe.labels_path, header=None)
norm = raw.astype(str).apply(lambda col: col.str.strip().str.lower())
hdr_row = norm.index[(norm == "condition").any(axis=1)][0]

cols = ["case", "spacecraft", "condition", "SV1", "SV2", "SV3", "SV4",
        "BP1", "BP2", "BP3", "BP4", "BP5", "BP6", "BP7", "BV1"]
lab_sv = raw.loc[hdr_row + 1:, :].iloc[:, :15].copy()
lab_sv.columns = cols
lab_sv["case"] = pd.to_numeric(lab_sv["case"], errors="coerce")
lab_sv = lab_sv.dropna(subset=["case"]).copy()
lab_sv["case"] = lab_sv["case"].astype(int)
for c in VALVE_NAMES:
    lab_sv[c] = pd.to_numeric(lab_sv[c], errors="coerce")

sv = lab_sv[lab_sv["condition"].astype(str).str.strip().str.lower() == "fault"].copy()

# failed valve = the one whose opening ratio is below 100
failed = sv[VALVE_NAMES].lt(100)
sv["n_failed"] = failed.sum(axis=1)
print("SV-fault recordings:", len(sv))
print("Failed valves per recording:\n", sv["n_failed"].value_counts().to_string())
assert (sv["n_failed"] == 1).all(), "some fault recordings have 0 or several failed valves"
sv["valve"] = failed.idxmax(axis=1)

print("\nRecordings per failed valve:\n", sv["valve"].value_counts().sort_index().to_string())
print("\nOpening ratios seen per valve:")
display(pd.crosstab(sv["valve"], sv.apply(lambda r: int(r[r["valve"]]), axis=1)))

# recording -> failed valve index (0..3)
valve_of = {int(c): VALVE_NAMES.index(v) for c, v in zip(sv["case"], sv["valve"])}
assert all(i in set(feat_df["recording_id"]) for i in valve_of), "some SV-fault recordings are not in feat_df"

# ---- Split by recording: 60 / 20 / 20, stratified by failed valve ----
ids4 = np.array(sorted(valve_of))
y4 = np.array([valve_of[i] for i in ids4])
ids_tr, ids_tmp, y_tr, y_tmp = train_test_split(ids4, y4, test_size=0.40, stratify=y4, random_state=SEED4)
ids_va, ids_te, y_va, y_te = train_test_split(ids_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=SEED4)
split4 = {"train": [int(i) for i in ids_tr], "val": [int(i) for i in ids_va], "test": [int(i) for i in ids_te]}

assert not (set(split4["train"]) & set(split4["val"]))
assert not (set(split4["train"]) & set(split4["test"]))
assert not (set(split4["val"])   & set(split4["test"]))
with open("split_valve.json", "w") as f:
    json.dump(split4, f)

print("\nRecordings per set and failed valve:")
for s, ids_s in split4.items():
    cnt = np.bincount([valve_of[i] for i in ids_s], minlength=NCLS)
    print(f"  {s:5s}: {len(ids_s):2d} recordings | " + ", ".join(f"{n}={c}" for n, c in zip(VALVE_NAMES, cnt)))

# ---- Windows of each split, taken from feat_df ----
F4 = {}
for s in split4:
    d = feat_df[feat_df["recording_id"].isin(split4[s])].copy()
    d["label"] = d["recording_id"].map(valve_of).astype(int)        # label is now the failed valve
    F4[s] = d.reset_index(drop=True)

# drop constant features (judged on train), standardize (fit on train only)
fcols4 = [c for c in feat_df.columns if c not in META_COLS]
keep4 = F4["train"][fcols4].to_numpy(float).std(axis=0) > 1e-12
fcols4 = [c for c, k in zip(fcols4, keep4) if k]
scaler4 = StandardScaler().fit(F4["train"][fcols4])
X4    = {s: scaler4.transform(F4[s][fcols4]) for s in F4}
meta4 = {s: F4[s][["recording_id", "window", "label"]] for s in F4}
ytr4  = F4["train"]["label"].to_numpy()

print(f"\n{len(fcols4)} features | windows: " + ", ".join(f"{s}={len(F4[s])}" for s in F4))

# ---- Evaluation: recording level (majority vote of its windows), macro over the 4 valves ----
def eval_sv(meta, pred):
    m = meta[["recording_id", "label"]].copy()
    m["pred"] = np.asarray(pred).astype(int)
    rec = m.groupby("recording_id").agg(
        label=("label", "first"),
        vote=("pred", lambda p: int(np.bincount(p, minlength=NCLS).argmax())))
    yt, yp = rec["label"].to_numpy(), rec["vote"].to_numpy()
    L = list(range(NCLS))
    return {"rec": {
        "acc":       accuracy_score(yt, yp),
        "f1":        f1_score(yt, yp, labels=L, average="macro", zero_division=0),
        "precision": precision_score(yt, yp, labels=L, average="macro", zero_division=0),
        "recall":    recall_score(yt, yp, labels=L, average="macro", zero_division=0),
        "cm":        confusion_matrix(yt, yp, labels=L),     # rows = true, cols = predicted
        "n":         len(yt),
    }}

# ---- Fit the supervised classifiers ----
models4 = make_supervised()
models4["NaiveBayes"] = GaussianNB(priors=[1 / NCLS] * NCLS)      # equal priors, 4 classes
results4 = {}
for name, clf in models4.items():
    clf.fit(X4["train"], ytr4)
    results4[name] = {s: eval_sv(meta4[s], clf.predict(X4[s])) for s in ["val", "test"]}

# ---- Results table (same layout as Task 1) ----
print(f"\n===== {TASK4_TITLE} =====")
print("Recording level, F1 / precision / recall are macro averages over the 4 valves "
      "(precision and recall: test only):")
display(metrics_table(results4))

# ---- 4x4 confusion matrices (recording level) ----
def plot_cm4(res, split_name, ncol=4):
    names = list(res.keys()); nrow = int(np.ceil(len(names) / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(4.2 * ncol, 4.0 * nrow))
    axes = np.atleast_1d(axes).ravel()
    for ax, nm in zip(axes, names):
        r = res[nm][split_name]["rec"]; cm = r["cm"]
        ax.imshow(cm, cmap="Blues", vmin=0)
        for i in range(NCLS):
            for j in range(NCLS):
                ax.text(j, i, int(cm[i, j]), ha="center", va="center", fontsize=13,
                        color="white" if cm.max() and cm[i, j] > cm.max() / 2 else "black")
        ax.set_xticks(range(NCLS)); ax.set_xticklabels(VALVE_NAMES)
        ax.set_yticks(range(NCLS)); ax.set_yticklabels(VALVE_NAMES)
        ax.set_xlabel("Predicted"); ax.set_ylabel("True")
        ax.set_title(f"{nm}\nAcc={100*r['acc']:.1f}%  F1={r['f1']:.3f}", fontsize=10)
    for ax in axes[len(names):]:
        ax.axis("off")
    fig.suptitle(f"{TASK4_TITLE} - confusion matrices ({split_name}, recording level)", fontsize=13)
    plt.tight_layout(); plt.show()

plot_cm4(results4, "val")
plot_cm4(results4, "test")
```

### Task-5

```python
# ------------------------------------------------------------------
# Task 5: opening ratio of the failed valve (regression, snapped to the known levels)
# Needs from earlier cells: pipe, feat_df, META_COLS (and the Task 1 imports)
# ------------------------------------------------------------------
from sklearn.linear_model import Ridge, LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, r2_score

TASK5_TITLE = "Task 5: Opening ratio of the failed valve (SV-fault recordings)"
VALVE_NAMES = ["SV1", "SV2", "SV3", "SV4"]
SEED5 = 42

# ---- SV-fault table: failed valve and its opening ratio ----
raw = pd.read_excel(pipe.labels_path, header=None)
norm = raw.astype(str).apply(lambda col: col.str.strip().str.lower())
hdr_row = norm.index[(norm == "condition").any(axis=1)][0]

cols = ["case", "spacecraft", "condition", "SV1", "SV2", "SV3", "SV4",
        "BP1", "BP2", "BP3", "BP4", "BP5", "BP6", "BP7", "BV1"]
lab5 = raw.loc[hdr_row + 1:, :].iloc[:, :15].copy()
lab5.columns = cols
lab5["case"] = pd.to_numeric(lab5["case"], errors="coerce")
lab5 = lab5.dropna(subset=["case"]).copy()
lab5["case"] = lab5["case"].astype(int)
for c in VALVE_NAMES:
    lab5[c] = pd.to_numeric(lab5[c], errors="coerce")

sv5 = lab5[lab5["condition"].astype(str).str.strip().str.lower() == "fault"].copy()
failed5 = sv5[VALVE_NAMES].lt(100)
assert (failed5.sum(axis=1) == 1).all(), "some fault recordings have 0 or several failed valves"
sv5["valve"] = failed5.idxmax(axis=1)
sv5["ratio"] = sv5[VALVE_NAMES].where(failed5).min(axis=1).astype(int)

LEVELS = [int(x) for x in sorted(sv5["ratio"].unique())]     # e.g. [0, 25, 50, 75]
print("SV-fault recordings:", len(sv5), "| opening-ratio levels:", LEVELS)
print("\nRecordings per failed valve and opening ratio:")
display(pd.crosstab(sv5["valve"], sv5["ratio"]))

ratio_of = {int(c): int(r) for c, r in zip(sv5["case"], sv5["ratio"])}
valve_name_of = {int(c): v for c, v in zip(sv5["case"], sv5["valve"])}
assert all(i in set(feat_df["recording_id"]) for i in ratio_of), "some SV-fault recordings are not in feat_df"

# ---- Split by recording: 70 / 30 (train / test), stratified by opening ratio ----
ids5 = np.array(sorted(ratio_of))
y5 = np.array([ratio_of[i] for i in ids5])
ids_tr, ids_te = train_test_split(ids5, test_size=0.30, stratify=y5, random_state=SEED5)
split5 = {"train": [int(i) for i in ids_tr], "test": [int(i) for i in ids_te]}

assert not (set(split5["train"]) & set(split5["test"]))
with open("split_ratio.json", "w") as f:
    json.dump(split5, f)

print("\nRecordings per set and opening ratio:")
for s, ids_s in split5.items():
    cnt = [sum(ratio_of[i] == L for i in ids_s) for L in LEVELS]
    print(f"  {s:5s}: {len(ids_s):2d} recordings | " + ", ".join(f"{L}%={c}" for L, c in zip(LEVELS, cnt)))

# ---- Windows from feat_df + failed valve (one-hot) as extra inputs ----
F5 = {}
for s in split5:
    d = feat_df[feat_df["recording_id"].isin(split5[s])].copy()
    d["label"] = d["recording_id"].map(ratio_of).astype(int)        # label is now the opening ratio
    for v in VALVE_NAMES:
        d[f"valve_{v}"] = (d["recording_id"].map(valve_name_of) == v).astype(float)
    F5[s] = d.reset_index(drop=True)

fcols5 = [c for c in F5["train"].columns if c not in META_COLS]
keep5 = F5["train"][fcols5].to_numpy(float).std(axis=0) > 1e-12
fcols5 = [c for c, k in zip(fcols5, keep5) if k]
scaler5 = StandardScaler().fit(F5["train"][fcols5])
X5    = {s: scaler5.transform(F5[s][fcols5]) for s in F5}
meta5 = {s: F5[s][["recording_id", "window", "label"]] for s in F5}
ytr5  = F5["train"]["label"].to_numpy(float)
print(f"\n{len(fcols5)} inputs (incl. valve one-hot) | windows: " +
      ", ".join(f"{s}={len(F5[s])}" for s in F5))

# ---- Regressors (target scaled to 0-1) ----
def make_regressors():
    return {
        "LinearRegression": LinearRegression(),
        "Ridge":            Ridge(alpha=10.0),
        "DecisionTree":     DecisionTreeRegressor(max_depth=5, random_state=0),
    }

LV = np.array(LEVELS)
def snap(v):
    """Round predictions to the nearest known opening ratio."""
    return LV[np.abs(np.asarray(v)[:, None] - LV[None, :]).argmin(axis=1)]

def eval_ratio(meta, pred_pct):
    """Recording level: mean of its windows, snapped to the nearest level for acc / F1 / precision / recall."""
    m = meta[["recording_id", "label"]].copy()
    m["pred"] = np.clip(pred_pct, LEVELS[0], LEVELS[-1])
    rec = m.groupby("recording_id").agg(label=("label", "first"), pred=("pred", "mean"))
    yt = rec["label"].to_numpy(int); yp = rec["pred"].to_numpy(float); ys = snap(yp)
    return {"rec": {
        "acc":       accuracy_score(yt, ys),
        "f1":        f1_score(yt, ys, labels=LEVELS, average="macro", zero_division=0),
        "precision": precision_score(yt, ys, labels=LEVELS, average="macro", zero_division=0),
        "recall":    recall_score(yt, ys, labels=LEVELS, average="macro", zero_division=0),
        "MAE":       mean_absolute_error(yt, yp),
        "R2":        r2_score(yt, yp),
        "cm":        confusion_matrix(yt, ys, labels=LEVELS),      # rows = true, cols = predicted
        "n":         len(yt),
    }}

results5 = {}
for name, reg in make_regressors().items():
    reg.fit(X5["train"], ytr5 / 100.0)
    results5[name] = {"test": eval_ratio(meta5["test"], 100.0 * reg.predict(X5["test"]))}

# reference: always predict the training mean
mean_pred = float(ytr5.mean())
results5["Baseline (train mean)"] = {"test": eval_ratio(meta5["test"], np.full(len(F5["test"]), mean_pred))}

# ---- Results table (test set, recording level) ----
tab5 = pd.DataFrame([{
    "model":          n,
    "type":           "baseline" if n.startswith("Baseline") else "regression",
    "test_acc%":      round(100 * r["test"]["rec"]["acc"], 1),
    "test_F1":        round(r["test"]["rec"]["f1"], 3),
    "test_precision": round(r["test"]["rec"]["precision"], 3),
    "test_recall":    round(r["test"]["rec"]["recall"], 3),
} for n, r in results5.items()]).set_index("model")
print(f"\n===== {TASK5_TITLE} =====")
print("Test set, recording level (mean of its windows, snapped to the nearest level); "
      "F1 / precision / recall are macro averages over the levels:")
display(tab5)

# ---- Confusion matrices (snapped predictions, recording level) ----
def plot_cm5(res, split_name, ncol=4):
    names = list(res.keys()); nrow = int(np.ceil(len(names) / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(4.2 * ncol, 4.0 * nrow))
    axes = np.atleast_1d(axes).ravel()
    k = len(LEVELS)
    for ax, nm in zip(axes, names):
        r = res[nm][split_name]["rec"]; cm = r["cm"]
        ax.imshow(cm, cmap="Blues", vmin=0)
        for i in range(k):
            for j in range(k):
                ax.text(j, i, int(cm[i, j]), ha="center", va="center", fontsize=13,
                        color="white" if cm.max() and cm[i, j] > cm.max() / 2 else "black")
        ax.set_xticks(range(k)); ax.set_xticklabels([f"{L}%" for L in LEVELS])
        ax.set_yticks(range(k)); ax.set_yticklabels([f"{L}%" for L in LEVELS])
        ax.set_xlabel("Predicted"); ax.set_ylabel("True")
        ax.set_title(f"{nm}\nAcc={100*r['acc']:.1f}%  F1={r['f1']:.3f}  MAE={r['MAE']:.1f}", fontsize=10)
    for ax in axes[len(names):]:
        ax.axis("off")
    fig.suptitle(f"{TASK5_TITLE} - confusion matrices ({split_name}, recording level)", fontsize=13)
    plt.tight_layout(); plt.show()

plot_cm5(results5, "test")
```

```python
# ------------------------------------------------------------------
# Save the trained models (run after the Task 1-5 cells)
#   Tasks 1-4: ANN, GradBoost, LogReg      Task 5: Ridge
# Each task is saved as one file with its models + the scaler + the feature list it was trained on.
# ------------------------------------------------------------------
import os, json, joblib

MODEL_DIR = "saved_models"
os.makedirs(MODEL_DIR, exist_ok=True)
KEEP = ["ANN", "GradBoost", "LogReg"]

def fit_and_check(task, Xtr, ytr, Xte, meta_te, res, eval_fn):
    """Refit the chosen models on the task's training set (same settings and seeds as in the
    task cell) and check that each one reproduces the test accuracy shown in the results table."""
    base, models = make_supervised(), {}
    for name in KEEP:
        clf = base[name].fit(Xtr, ytr)
        acc = eval_fn(meta_te, clf.predict(Xte))["rec"]["acc"]
        assert np.isclose(acc, res[name]["test"]["rec"]["acc"]), \
            f"{task} {name}: refitted model does not match the table ({acc:.3f})"
        models[name] = clf
    return models

bundles = {
    "task1": {"models": fit_and_check("Task 1", X["train"], ytr, X["test"], meta["test"], results, evaluate),
              "scaler": scaler, "fcols": fcols, "classes": ["Normal", "Abnormal"]},
    "task2": {"models": fit_and_check("Task 2", X2["train"], ytr2, X2["test"], meta2["test"], results2, evaluate),
              "scaler": scaler2, "fcols": fcols2, "classes": TYPE_NAMES},
    "task3": {"models": fit_and_check("Task 3", X3["train"], ytr3, X3["test"], meta3["test"], results3, eval_loc),
              "scaler": scaler3, "fcols": fcols3, "classes": LOC_NAMES},
    "task4": {"models": fit_and_check("Task 4", X4["train"], ytr4, X4["test"], meta4["test"], results4, eval_sv),
              "scaler": scaler4, "fcols": fcols4, "classes": VALVE_NAMES},
}

# Task 5: Ridge regression (target scaled to 0-1 during training)
ridge = make_regressors()["Ridge"].fit(X5["train"], ytr5 / 100.0)
acc5 = eval_ratio(meta5["test"], 100.0 * ridge.predict(X5["test"]))["rec"]["acc"]
assert np.isclose(acc5, results5["Ridge"]["test"]["rec"]["acc"]), "Task 5 Ridge: refitted model does not match the table"
bundles["task5"] = {"models": {"Ridge": ridge}, "scaler": scaler5, "fcols": fcols5,
                    "levels": LEVELS, "valve_names": VALVE_NAMES, "target_scale": 100.0}

for task, b in bundles.items():
    joblib.dump(b, os.path.join(MODEL_DIR, f"{task}.joblib"))

# feature settings, so new data gets exactly the same features
feature_settings = {"windows": [list(w) for w in pipe.windows_def], "sensors": list(pipe.sensors),
                    "fs": pipe.fs, "n_rows": pipe.n_rows, "drop_features": sorted(pipe.drop_features)}
with open(os.path.join(MODEL_DIR, "feature_settings.json"), "w") as f:
    json.dump(feature_settings, f, indent=2)

# ---- what was saved ----
res_of = {"task1": results, "task2": results2, "task3": results3, "task4": results4, "task5": results5}
rows = [{"task": t, "model": m, "test_acc%": round(100 * res_of[t][m]["test"]["rec"]["acc"], 1),
         "test_F1": round(res_of[t][m]["test"]["rec"]["f1"], 3), "file": f"{MODEL_DIR}/{t}.joblib"}
        for t, b in bundles.items() for m in b["models"]]
print("Saved models (test results from the task cells):")
display(pd.DataFrame(rows).set_index(["task", "model"]))
print("Feature settings saved to:", os.path.abspath(os.path.join(MODEL_DIR, "feature_settings.json")))
```

```python
# ------------------------------------------------------------------
# Predict on NEW data with the saved models (no training here)
# Can run in a fresh kernel: needs only the saved_models folder and feature_engineering.py
#   Task 1 -> Normal / Abnormal
#   Task 2 -> abnormal recordings: Bubble / SV fault
#   Task 3 -> Bubble recordings: location (BP1-BP7, BV1)
#   Task 4 -> SV-fault recordings: failed valve (SV1-SV4)
#   Task 5 -> SV-fault recordings: opening ratio of that valve (uses the Task 4 valve)
# ------------------------------------------------------------------
import os, re, glob, json, joblib
import numpy as np
import pandas as pd
from feature_engineering import FeatureEngineering

MODEL_DIR    = "saved_models"
NEW_DATA_DIR = test_data                     # folder with the new CaseXXX.csv files
OUT_CSV      = "predictions_new_data.csv"
USE = {"task1": "ANN", "task2": "ANN", "task3": "ANN", "task4": "ANN"}   # or "GradBoost" / "LogReg"

# ---- load the saved models and feature settings ----
B = {t: joblib.load(os.path.join(MODEL_DIR, f"{t}.joblib")) for t in ["task1", "task2", "task3", "task4", "task5"]}
with open(os.path.join(MODEL_DIR, "feature_settings.json")) as f:
    FEAT_SET = json.load(f)
fe = FeatureEngineering(NEW_DATA_DIR, "", windows=[tuple(w) for w in FEAT_SET["windows"]],
                        sensors=FEAT_SET["sensors"], fs=FEAT_SET["fs"], n_rows=FEAT_SET["n_rows"],
                        drop_features=FEAT_SET["drop_features"])

# ---- window features of the new recordings (same as training, no labels needed) ----
def new_features(data_dir):
    files = sorted(glob.glob(os.path.join(data_dir, "*.csv")))
    assert files, f"no CSV files in {data_dir}"
    need = max(b for _, b in fe.windows_def)
    rows = []
    for f in files:
        rid = int(re.search(r"(\d+)", os.path.basename(f)).group(1))     # Case178.csv -> 178
        d = pd.read_csv(f)
        d.columns = [c.strip().upper() for c in d.columns]
        assert all(s in d.columns for s in fe.sensors), f"{f}: sensor columns missing"
        arr = d[fe.sensors].to_numpy(float)[:fe.n_rows]
        assert len(arr) >= need, f"{f}: only {len(arr)} rows, need {need}"
        for w, (a, b) in enumerate(fe.windows_def):
            row = {"recording_id": rid, "window": w, "file": os.path.basename(f)}
            row.update(fe.window_features(arr[a:b]))
            rows.append(row)
    df = pd.DataFrame(rows)
    fc = [c for c in df.columns if c not in ("recording_id", "window", "file")]
    df[fc] = np.nan_to_num(df[fc].to_numpy(float))
    return df

def vote(df, pred):
    """One label per recording: majority vote of its windows."""
    t = pd.DataFrame({"recording_id": df["recording_id"].to_numpy(), "pred": np.asarray(pred).astype(int)})
    return t.groupby("recording_id")["pred"].agg(lambda p: int(np.bincount(p).argmax()))

def classify(task, df):
    b = B[task]
    missing = [c for c in b["fcols"] if c not in df.columns]
    assert not missing, f"{task}: features missing in the new data, e.g. {missing[:3]}"
    Xn = b["scaler"].transform(df[b["fcols"]])
    return vote(df, b["models"][USE[task]].predict(Xn))

new_df = new_features(NEW_DATA_DIR)
print(f"New data: {new_df['recording_id'].nunique()} recordings, {len(new_df)} windows")

out = new_df.groupby("recording_id")[["file"]].first()
for c in ["condition", "fault_type", "bubble_location", "failed_valve", "opening_ratio_%"]:
    out[c] = pd.Series("-", index=out.index, dtype=object)

# ---- Task 1: normal / abnormal ----
t1 = classify("task1", new_df)
out["condition"] = t1.map(dict(enumerate(B["task1"]["classes"])))
abn = new_df[new_df["recording_id"].map(t1) == 1]

if len(abn):
    # ---- Task 2: Bubble / SV fault ----
    t2 = classify("task2", abn)
    out.loc[t2.index, "fault_type"] = t2.map(dict(enumerate(B["task2"]["classes"])))

    # ---- Task 3: bubble location ----
    bub = abn[abn["recording_id"].map(t2) == 0]
    if len(bub):
        t3 = classify("task3", bub)
        out.loc[t3.index, "bubble_location"] = t3.map(dict(enumerate(B["task3"]["classes"])))

    svf = abn[abn["recording_id"].map(t2) == 1]
    if len(svf):
        # ---- Task 4: failed valve ----
        t4 = classify("task4", svf)
        out.loc[t4.index, "failed_valve"] = t4.map(dict(enumerate(B["task4"]["classes"])))

        # ---- Task 5: opening ratio (valve one-hot taken from the Task 4 prediction) ----
        b5 = B["task5"]
        d5 = svf.copy()
        for i, v in enumerate(b5["valve_names"]):
            d5[f"valve_{v}"] = (d5["recording_id"].map(t4) == i).astype(float)
        pct = b5["target_scale"] * b5["models"]["Ridge"].predict(b5["scaler"].transform(d5[b5["fcols"]]))
        lv = np.array(b5["levels"])
        rec = pd.Series(np.clip(pct, lv[0], lv[-1]), index=d5["recording_id"].to_numpy()).groupby(level=0).mean()
        snapped = lv[np.abs(rec.to_numpy()[:, None] - lv[None, :]).argmin(axis=1)]
        out.loc[rec.index, "opening_ratio_%"] = snapped.astype(int)

out.to_csv(OUT_CSV)
print(f"Models used: Task 1-4 = {USE}, Task 5 = Ridge")
print("Predictions saved to:", os.path.abspath(OUT_CSV))
print("\nPredicted condition:", out["condition"].value_counts().to_dict())
display(out)
```
