```python
training_data = "C:/Users/Lakshya/Desktop/Data_Mining/Course_Project/Dataset/train/data/"
training_labels = "C:/Users/Lakshya/Desktop/Data_Mining/Course_Project/Dataset/train/labels.xlsx/"
test_data = "C:/Users/Lakshya/Desktop/Data_Mining/Course_Project/Dataset/test/data/"
test_labels = "C:/Users/Lakshya/Desktop/Data_Mining/Course_Project/Dataset/test/labels_spacecraft.xlsx/"
```

```python
from feature_engineering import FeatureEngineering
from fault_tasks import TaskTrainer
from fault_predictor import FaultPredictor
from dashboard import Dashboard
```

```python
# 1) features + train Tasks 1-5 + save the models
pipe = FeatureEngineering(training_data, training_labels)
pipe.run()
trainer = TaskTrainer(pipe, model_dir="saved_models").run_all()
trainer.show()                          # results table + test confusion matrices per task
```

```python
# 2) test folder -> Tasks 1-5 -> written into the test labels file
predictor = FaultPredictor("saved_models")        # ANN by default for Tasks 1-4, Ridge for Task 5
pred = predictor.run(test_data, test_labels)
```

```python
Dashboard(pipe=pipe, trainer=trainer, predictions=pred).build("dashboard.html")
```
